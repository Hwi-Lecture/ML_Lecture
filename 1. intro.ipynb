### 데이터 전처리

```python
X = X.drop(columns=['PassengerId', 'Name', 'Ticket', 'Cabin'])
X['Sex'] = X['Sex'].map({'male': 0, 'female': 1})
X['Embarked'] = X['Embarked'].map({'S': 1, 'C': 2, 'Q': 3})

X.head()
```

이제 문자열을 모두 처리했으니 학습이 가능할까요?

```python
model = LogisticRegression(random_state=42)
model.fit(X, y)
```

또 에러가 떴죠?

"Input X contains NaN."

이건 결측치(NaN)가 데이터에 포함되어 있어 계산이 불가능하다는 것을 의미합니다.

이런 식으로 결측치가 있으면 학습하지 못하는 경우가 많기 때문에 대부분 머신러닝 모델을 학습하기 전에는 결측치를 꼭 없애고 학습을 하는게 좋습니다. (DecisionTree와 같은 의사결정나무 모델은 결측치가 있어도 학습이 되기는 합니다.)

결측치의 처리 방법으로는,
* 행 삭제 : 결측치가 있는 승객 정보를 제외하는 방법.
* 열 제외 : 결측치가 있는 컬럼을 제외하는 방법.
* 값 채우기 : 해당 데이터를 유지하면서, 실제와 다른 값으로 채워 넣는다.

결측치를 어떻게 처리하느냐에 따라서도 머신러닝 모델의 결과가 많이 달라지게 됩니다.

```python
X.isna().sum()
```

우선 지금은 Age 컬럼은 Age 컬럼의 중앙값, Embarked 컬럼은 최빈값으로 채워 넣도록 하겠습니다.

```python
X['Age'] = X['Age'].fillna(X['Age'].median())
X['Embarked'] = X['Embarked'].fillna(X['Embarked'].mode()[0])

X.isna().sum()
```

이번에야 말로 학습을 해볼 수 있을까요?

```python
model = LogisticRegression(random_state=42)
model.fit(X, y)
```

드디어 학습이 되었습니다! 성능을 한번 확인해 볼까요? 정확도를 한번 측정해 보겠습니다.

```python
model.score(X, y)
```

80%의 정확도를 달성했습니다!

그래서 학습용 데이터와 평가용 데이터를 나눠서 모델 예측을 수행하게 됩니다.

학습용 데이터는 모델이 학습할 때 보는 데이터, 평가용 데이터는 모델이 학습할 때 보지 못한 데이터.

이렇게 나눠야 모델이 새로운 데이터를 만났을 때도 잘 맞추는지를 확인할 수 있는거죠.

```python
from sklearn.model_selection import train_test_split

df = pd.read_csv("Titanic-Dataset.csv")

y = df['Survived']
X = df.drop(columns=['Survived'])

X = X.drop(columns=['PassengerId', 'Name', 'Ticket', 'Cabin'])
X['Sex'] = X['Sex'].map({'male': 0, 'female': 1})
X['Embarked'] = X['Embarked'].map({'S': 1, 'C': 2, 'Q': 3})

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# 결측치를 처리할 때 사용했던 결측치 처리를 다시 나눠서 수행합니다.
X_train['Age'] = X_train['Age'].fillna(X_train['Age'].median())
X_train['Embarked'] = X_train['Embarked'].fillna(X_train['Embarked'].mode()[0])

X_test['Age'] = X_test['Age'].fillna(X_train['Age'].median())
X_test['Embarked'] = X_test['Embarked'].fillna(X_train['Embarked'].mode()[0])

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

다시, 훈련셋으로 모델을 훈련하고 테스트셋으로 모델 성능을 예측해 보겠습니다.

```python
model = LogisticRegression(random_state=42)
model.fit(X_train, y_train)

model.score(X_test, y_test)
```

처음 본 데이터에 대해서도 80%의 정확도를 달성하네요. 앞의 결과보다 좀 더 신빈성 있는 결과를 얻은 것 같습니다.
